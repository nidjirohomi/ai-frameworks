## Очистка

```python
df_clean = df.drop_duplicates().copy()
print("shape after drop_duplicates", df_clean.shape)
```

```python
tip_median = df_clean['tip'].median()
print("tip median: ", tip_median)

df_clean['tip'] = df_clean['tip'].fillna(tip_median)
```

```python
rating_median = df_clean['rating'].median()
print("rating median: ", rating_median)

df_clean['rating'] = df_clean['rating'].fillna(rating_median)
```

## 5. Анализы: фильтры и groupby

Несколько коротких вопросов к данным — по одной идее на ячейку.

### 5.1. Средний чек у членов программы лояльности и остальных


```python
result_5_1 = df_clean.groupby('is_member')['price'].mean()
print(result_5_1)
```

### 5.2. Какие напитки берут вечером и ночью?


```python
evening_night_drinks = df_clean[df_clean['time_slot'].isin(['evening', 'late_night'])]
print(evening_night_drinks['drink'].value_counts())
```

### 5.3. Среднее время приготовления по типу еды


```python
result_5_3 = df_clean.groupby('food_item')['prep_minutes'].mean()
print(result_5_3)
```

### 5.4. Дождь и горячий шоколад — совпадение?


```python
df_clean['has_hot_choco'] = df_clean['drink'] == 'hot_chocolate'
print(pd.crosstab(df_clean['raining'], df_clean['has_hot_choco'], normalize='index') * 100)
```

### 5.5. Курс студента и средний рейтинг / чаевые


```python
result_5_5 = df_clean.groupby('student_year')[['rating', 'tip']].mean()
print(result_5_5)
```

### 5.6. Факультет и калории (кто «слаще» заказывает?)


```python
result_5_6 = df_clean.groupby('faculty')['calories'].mean().sort_values(ascending=False)
print(result_5_6)
```

## 6. Графики

Подписываем оси и заголовки. Без лишнего декора. Каждую диаграмму следет построить через plt и seaborn


### 6.1. Гистограмма цены

```python
plt.figure(figsize=(8, 5))
sns.histplot(df_clean['price'], bins=20, kde=True, color='skyblue')
plt.title('Гистограмма цен заказов')
plt.xlabel('Цена')
plt.ylabel('Частота')
plt.show()
```

### 6.2. Столбчатая диаграмма: заказы по дням недели


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df_clean, x='day_of_week', order=['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'], palette='viridis')
plt.title('Колво заказов по дням недели')
plt.xlabel('День недели')
plt.ylabel('Колво заказов')
plt.show()
```

### 6.3. Средняя цена по напитку (barh удобнее читать)


```python
plt.figure(figsize=(8, 5))
avg_price_drink = df_clean.groupby('drink')['price'].mean().sort_values()
avg_price_drink.plot(kind='barh', color='coral')
plt.title('Средняя цена по напиткам')
plt.xlabel('Средняя цена')
plt.ylabel('Напиток')
plt.show()
```

### 6.4. Scatter: время приготовления vs рейтинг


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x='prep_minutes', y='rating', alpha=0.7, color='purple')
plt.title('Время приготовления vs Рейтинг')
plt.xlabel('Время приготовления (мин)')
plt.ylabel('Рейтинг')
plt.show()
```

### 6.5. Boxplot калорий по размеру стакана


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_clean, x='size', y='calories', order=['S', 'M', 'L'], palette='Set2')
plt.title('Калории в зависимости от размера стакана')
plt.xlabel('Размер')
plt.ylabel('Калории')
plt.show()
```

### 6.6. Связь учёбы и кофе: study hours vs prior cups (scatter + лёгкий цвет по member)

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x='study_hours_today', y='prior_coffee_cups', hue='is_member', palette='deep', alpha=0.8)
plt.title('Связь часов учебы и выпитого кофе до заказа')
plt.xlabel('Часы учебы сегодня')
plt.ylabel('Количество чашек кофе до этого')
plt.legend(title='Участник программы')
plt.show()
```

### Задание A

Посчитайте **средний `tip`** для каждого способа оплаты (`payment`).  
Отсортируйте по убыванию среднего чаевого. Постройте bar chart.


```python
avg_tip_payment = df_clean.groupby('payment')['tip'].mean().sort_values(ascending=False)
print(avg_tip_payment)

plt.figure(figsize=(8, 5))
avg_tip_payment.plot(kind='bar', color='teal')
plt.title('Средние чаевые по способам оплаты')
plt.xlabel('Способ оплаты')
plt.ylabel('Средний размер чаевых')
plt.xticks(rotation=0)
plt.show()
```
