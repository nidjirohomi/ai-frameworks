```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('campus-commute-late.csv')

df.head()
```

```python
df.info()
```

```python
df['phone_battery'] = df['phone_battery'].fillna(df['phone_battery'].median())
```

```python
df.info()
```

```python
df.describe()
```

```python
df['commute_mode'].value_counts()
```

```python
df.columns
```

```python
numeric_features = ['distance_km', 'leave_hour',
       'buffer_minutes', 'sleep_hours', 'raining', 'temperature_c',
       'traffic_index', 'backpack_kg', 'phone_battery', 'woke_late'
       ]
```

```python
y = df['is_late']
```

```python
X_num = df[numeric_features]
```

```python
X_num.head()
```

```python
X_mode = pd.get_dummies(df['commute_mode'], prefix= 'mode')
```

```python
X_mode.head()
```

```python
X = pd.concat([X_num, X_mode], axis=1)
```

```python
X.head()
```

```python
print(X.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

print("train:", X_train.shape, "test:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

pred = knn.predict(X_test)

print(pred)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, pred)

print("k=5 accuracy WITHOUT scaling:", accuracy)
```

```python
scaler = StandardScaler()

X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.fit_transform(X_test)

knn = KNeighborsClassifier()
knn.fit(X_train_s, y_train)

pred = knn.predict(X_test_s)

accuracy = accuracy_score(y_test, pred)

print("k=5 accuracy WITH scaling:", accuracy)
```

```python
X_train_s[:10]
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
cm = confusion_matrix(y_test, pred)
print(cm)

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['on_time(0)', 'late(1)'])
disp.plot(cmap='Blues')
plt.title('k-NN (k=5, scaled)')
plt.show()
```

1. Переберите несколько значений k и сравните accuracy на тесте. Постройте график k и accuracy которая ей соответствует

```python
k_values = range(1, 21)
accuracies = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_s, y_train)
    pred = knn.predict(X_test_s)
    acc = accuracy_score(y_test, pred)
    accuracies.append(acc)

plt.figure(figsize=(8, 5))
plt.plot(k_values, accuracies, marker='o', linestyle='-', color='b')
plt.title('Зависимость Accuracy от значения k')
plt.xlabel('Значение k')
plt.xticks(k_values)
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

```python
X_day = pd.get_dummies(df['day_of_week'], prefix='day')
X_full = pd.concat([X_num, X_mode, X_day], axis=1)
X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(
    X_full, y, test_size=0.25, random_state=42, stratify=y
)


scaler_a = StandardScaler()
X_train_a_s = scaler_a.fit_transform(X_train_a)
X_test_a_s = scaler_a.transform(X_test_a)
knn_a = KNeighborsClassifier(n_neighbors=5)
knn_a.fit(X_train_a_s, y_train_a)

pred_a = knn_a.predict(X_test_a_s)
accuracy_a = accuracy_score(y_test_a, pred_a)
print("Accuracy с учетом day_of_week (k=5):", accuracy_a)
```
