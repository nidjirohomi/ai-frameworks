```python
def mean(values):
  total = 0
  for v in values:
    total = total + v
  return total / len(values)

print(mean([10, 20, 30]))
```

```python
def my_min(numbers):
    current_min = numbers[0]
    for num in numbers:
        if num < current_min:
            current_min = num
    return current_min

def my_max(numbers):
    current_max = numbers[0]
    for num in numbers:
        if num > current_max:
            current_max = num
    return current_max


data = [15, 42, 8, 23]
print(my_min(data))
print(my_max(data))
```

```python
def my_mean(values):
    total = 0
    for v in values:
        total = total + v
    return total / len(values)

def my_min(values):
    minimum = values[0]
    for v in values[1:]:
        if v < minimum:
            minimum = v
    return minimum

def my_max(values):
    maximum = values[0]
    for v in values[1:]:
        if v > maximum:
            maximum = v
    return maximum

data = [5, 2, 9, 2, 7]
print(my_mean(data), my_min(data), my_max(data))
```

```python
import csv
with open("tiny.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerow({"name": "Ada", "score": 90})
    writer.writerow({"name": "Alan", "score": 80})
    writer.writerow({"name": "Grace", "score": 95})
scores = []
with open("tiny.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        scores.append(float(row["score"]))

print(scores)
print(my_mean(scores))
```
